```python
toy3 = pd.DataFrame({"x": [3, 0, 1], "y": [4, 0, 2], "z": [0, 5, 2]})
toy_mag = np.sqrt(toy3["x"] ** 2 + toy3["y"] ** 2 + toy3["z"] ** 2)
toy_mag
```
